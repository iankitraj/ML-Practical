```python
def highlight_petal_length(s):
    '''
    Highlight petal length greater than 6 for virginica species.
    '''
    is_virginica = s.Species == 'virginica'
    is_long_petal = s.Petal_Length > 6
    return ['background-color: yellow' if is_virginica and is_long_petal else '' for v in s]

styled_iris = iris.style.apply(highlight_petal_length, axis=1)  # Apply styling row-wise
styled_iris  # Display the styled DataFrame
```
