Q. 1 Build Decision Tree using sklearn with given dataset.

Download and read dataset users.csv Import necessary files to implement
decision tree classification. Define predictors and target. Perform one hot
encoding with required columns. Implement decision tree classification.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score
```

```python
user = pd.read_csv('users.csv')
```

```python
user
```

```python
user.info()
```

```python
user.head()
```

```python
x = user.columns.drop('Purchased')
y = user["Purchased"]

user_encoded = pd.get_dummies(user[x]).astype(int)

user_encoded
```

```python
x_train, x_test, y_train, y_test = train_test_split(user_encoded, y, test_size = 0.15, random_state = 0)
```

```python
model = DecisionTreeClassifier()
model.fit(x_train, y_train)
```

```python
train_prediction = model.predict(x_train)
test_prediction = model.predict(x_test)
```

```python
plt.figure(figsize=(30, 30))
plot_tree(model, feature_names=user_encoded.columns, filled=True, rounded=True, fontsize=15, class_names=['Not purchased', 'Purchased'])
plt.title("Decision tree Visualization")
plt.show()
```

Q. 3 Apply suitable Regression.
    
Read computer.csv dataset. Find no. of columns and rows with given
dataset. Which regression model fit with given dataset? Build suitable
regression model with the data and fetch intercept and coefficient. Visualize
Models using pyplot. Compute the Coefficient of Determination to find
accuracy

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
computers = pd.read_csv("computers .csv")
```

```python
computers
```

```python
mean_value = computers['Minutes'].mean()
print(mean_value)
```

```python
plt.scatter(computers['Units'],computers['Minutes'],color='blue', marker='o')
plt.axhline(y=mean_value,c='r')
plt.xlabel("Units")
plt.ylabel("Minutes")
```

```python
minutes_model0 = computers['Minutes'].mean()
minutes_model1 = 10 + 12*computers['Units'] 
minutes_model2 = 6 + 18*computers['Units']  
```

```python
computers['min_model0'] = minutes_model0
computers['min_model1'] = minutes_model1
computers['min_model2'] = minutes_model2
computers
```

```python
fig,ax = plt.subplots()
ax.scatter(x= 'Units' , y="Minutes" , data = computers , label = 'Actual Repair Time')
ax.plot(computers['Units'], computers['min_model0'], color = 'red' , label = 'Model0')
ax.plot(computers['Units'], computers['min_model1'],  color = 'green' , label = 'Model1')
ax.plot(computers['Units'], computers['min_model2'],  color = 'black' , label = 'Model2')
ax.set_xlabel('Units')
ax.set_ylabel('Minutes')
ax.set_title('Speculated Models')
ax.legend()
```

```python
x = computers[['Units']]
y = computers['Minutes']
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x,y)
print("Intercept" ,model.intercept_)
print("Coefficient", model.coef_)
```

```python
minutes_model3 = 4.161654135338296 + 15.50877193*computers['Units']
computers['min_model3'] = minutes_model3
computers
```

```python
fig,ax = plt.subplots()
ax.scatter(x= 'Units' , y="Minutes" , data = computers , label = 'Actual Repair Time')
ax.plot(computers['Units'], computers['min_model0'], color = 'red' , label = 'Model0')
ax.plot(computers['Units'], computers['min_model1'],  color = 'green' , label = 'Model1')
ax.plot(computers['Units'], computers['min_model2'],  color = 'black' , label = 'Model2')
ax.plot(computers['Units'], computers['min_model3'],  color = 'yellow' , label = 'Model3')
ax.set_xlabel('Units')
ax.set_ylabel('Minutes')
ax.set_title('Speculated Models')
ax.legend()
```

```python
Rsq = model.score(computers[['Units']],y) * 100
Rsq
```
