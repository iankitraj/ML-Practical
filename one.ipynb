```python
#9(A) : Read the “mobile.csv” dataset.
df5 = pd.read_csv('mobile.csv')
df5
```

```python
#9(B) : Draw scatter plot of purchased and not_purchased value with respect to Age and EstimatedSalary.

purchased = df5[df5['Purchased'] == 1]
not_purchased = df5[df5['Purchased'] == 0]

plt.figure(figsize=(10, 6))
plt.scatter(purchased['Age'], purchased['EstimatedSalary'], color='green', label='Purchased')
plt.scatter(not_purchased['Age'], not_purchased['EstimatedSalary'], color='red', label='Not Purchased')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.title('Scatter Plot of Purchased and Not Purchased with respect to Age and Estimated Salary')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#9(C) : Do Feature engineering (normalize the Age and EstimatedSalary columns).
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df5[['Age', 'EstimatedSalary']] = scaler.fit_transform(df5[['Age', 'EstimatedSalary']])
```

```python
#10(A) : Split the dataset into train and test data (with reference of que. 9)
X = df5[['Age', 'EstimatedSalary']]
y = df5['Purchased']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=100)
```

```python
#10(B) : Build the kNN model with k=5,k=3,k=7.
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score


X = df5[['Age', 'EstimatedSalary']]
y = df5['Purchased']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

k_values = [5, 3, 7]

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"kNN model with k={k}: Accuracy = {accuracy:.2f}")
```

```python
#10(C) : Evaluate model performance on train and test sets with different values of k.
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score


X = df5[['Age', 'EstimatedSalary']]
y = df5['Purchased']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

k_values = [5, 3, 7]

for k in k_values:
    
    knn = KNeighborsClassifier(n_neighbors=k)
    
    knn.fit(X_train, y_train)
    
    y_train_pred = knn.predict(X_train)
    
    y_test_pred = knn.predict(X_test)
    
    train_accuracy = accuracy_score(y_train, y_train_pred)
    
    test_accuracy = accuracy_score(y_test, y_test_pred)
    print(f"kNN model with k={k}: Train Accuracy = {train_accuracy:.2f}, Test Accuracy = {test_accuracy:.2f}")
```
