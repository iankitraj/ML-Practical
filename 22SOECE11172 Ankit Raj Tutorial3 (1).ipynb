**7.Find categories of ‘source_city’. (categorical data)**

```python
df = pd.read_csv('/content/drive/MyDrive/AIML/Clean_Dataset.csv')
df = df.rename(columns={'Unnamed: 0':'id'})
print(df['source_city'].unique())
```

**8.Find categories of ‘class’. (categorical data)**

```python
df = pd.read_csv('/content/drive/MyDrive/AIML/Clean_Dataset.csv')
df = df.rename(columns={'Unnamed: 0':'id'})
print(df['class'].unique())
```

**9.Draw Count plot for ‘source_city’.**

```python
plt.figure(figsize=(15,6))
sns.countplot(x='source_city', data=df)
plt.show()
```

**10.Draw Count plot for ‘class’.**

```python
plt.figure(figsize=(15,6))
sns.countplot(x='class', data=df)
plt.show()
```

**11.Draw a pie plot for ‘source_city’.**

```python
plt.figure(figsize=(15,6))
df['source_city'].value_counts().plot.pie(autopct='%1.1f%%')
plt.show()
```

**12.Draw Donut plot for ‘source_city’.**


```python
plt.figure(figsize=(15,6))
df['source_city'].value_counts().plot.pie(autopct='%1.1f%%', wedgeprops=dict(width=0.3))
plt.show()
```

**13.Draw Histogram plot for ‘price’.**

```python
plt.figure(figsize=(15,6))
sns.histplot(df['price'])
plt.show()
```

**14.Draw Distplot for ‘price’**

```python
sns.displot(df['price'])
plt.show()
```
